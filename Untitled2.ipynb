```python
# ==============================================================================
# Task 1a: Count the number of images in each class (NORMAL vs PNEUMONIA)
# ==============================================================================

import os

# Updated base paths according to your Google Drive structure
train_dir = '/content/drive/MyDrive/archive (3)/chest_xray/train/'
test_dir = '/content/drive/MyDrive/archive (3)/chest_xray/test/'
val_dir = '/content/drive/MyDrive/archive (3)/chest_xray/val/'

# Print the number of images in the Training set
print("--- Training Set ---")
print("NORMAL images:", len(os.listdir(os.path.join(train_dir, 'NORMAL'))))
print("PNEUMONIA images:", len(os.listdir(os.path.join(train_dir, 'PNEUMONIA'))))

# Print the number of images in the Testing set
print("\n--- Testing Set ---")
print("NORMAL images:", len(os.listdir(os.path.join(test_dir, 'NORMAL'))))
print("PNEUMONIA images:", len(os.listdir(os.path.join(test_dir, 'PNEUMONIA'))))

# Print the number of images in the Validation set
print("\n--- Validation Set ---")
print("NORMAL images:", len(os.listdir(os.path.join(val_dir, 'NORMAL'))))
print("PNEUMONIA images:", len(os.listdir(os.path.join(val_dir, 'PNEUMONIA'))))
```

```python

import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# 1. Define standard image size and batch size for the model
IMG_HEIGHT = 150
IMG_WIDTH = 150
BATCH_SIZE = 32

# 2. Initialize ImageDataGenerator for rescaling (normalizing pixel values to 0-1)
# We apply rescaling to train, validation, and test datasets
train_datagen = ImageDataGenerator(rescale=1./255)
val_datagen = ImageDataGenerator(rescale=1./255)
test_datagen = ImageDataGenerator(rescale=1./255)

# 3. Load and prepare images from the directories
print(" Loading Training Data...")
train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(IMG_HEIGHT, IMG_WIDTH), # Resize all images to 150x150
    batch_size=BATCH_SIZE,
    class_mode='binary' # Binary classification (NORMAL vs PNEUMONIA)
)

print("\n Loading Validation Data...")
val_generator = val_datagen.flow_from_directory(
    val_dir,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='binary'
)

print("\n Loading Testing Data...")
test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode='binary',
    shuffle=False # Shuffle is False for test data to evaluate correctly later
)
```

```python

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout

# 1. Initialize the Sequential model
model = Sequential()

# 2. Add Convolutional and Pooling layers to extract features from images
model.add(Conv2D(32, (3, 3), activation='relu', input_shape=(150, 150, 3)))
model.add(MaxPooling2D(pool_size=(2, 2)))

model.add(Conv2D(64, (3, 3), activation='relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))

model.add(Conv2D(128, (3, 3), activation='relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))

# 3. Flatten the layers to feed into the Dense (fully connected) layers
model.add(Flatten())

# 4. Add Dense layers for classification
model.add(Dense(128, activation='relu'))

# Note: Dropout layer added as a solution to prevent Overfitting
model.add(Dropout(0.5))

# 5. Output layer: 1 neuron with 'sigmoid' for binary classification (Normal vs Pneumonia)
model.add(Dense(1, activation='sigmoid'))

# 6. Compile the model
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Print the model architecture summary
print(" Proposed CNN Model Architecture Built Successfully!\n")
model.summary()
```

```python
# ==============================================================================
# Task 1a: Training the Proposed CNN Model
# ==============================================================================

# Define the number of epochs (training rounds)
# 10 epochs is a good balance for initial training and performance evaluation
EPOCHS = 10

print("🚀 Starting Model Training... Please wait until all epochs are completed.\n")

# Train the model and save the progress in 'history' variable for later analysis
history = model.fit(
    train_generator,
    epochs=EPOCHS,
    validation_data=val_generator
)

print("\n🎉 Proposed CNN Model Training Completed Successfully!")
```

```python
# ==============================================================================
# Task 1c (vii): Result Evaluation - Performance Visualization
# ==============================================================================
import matplotlib.pyplot as plt

# Retrieve accuracy and loss values from the training history
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']
loss = history.history['loss']
val_loss = history.history['val_loss']
epochs_range = range(len(acc))

# Plotting Accuracy
plt.figure(figsize=(8, 4))
plt.plot(epochs_range, acc, label='Training Accuracy')
plt.plot(epochs_range, val_acc, label='Validation Accuracy')
plt.legend(loc='lower right')
plt.title('Training and Validation Accuracy')
plt.show()

# Plotting Loss
plt.figure(figsize=(8, 4))
plt.plot(epochs_range, loss, label='Training Loss')
plt.plot(epochs_range, val_loss, label='Validation Loss')
plt.legend(loc='upper right')
plt.title('Training and Validation Loss')
plt.show()

print("✅ Visualization complete! These graphs are essential for your report (Task 1c - vii).")
```

```python
# Evaluate the model on test data
test_loss, test_acc = model.evaluate(test_generator)
print(f"\n✅ Test Accuracy: {test_acc*100:.2f}%")
print(f"✅ Test Loss: {test_loss:.4f}")
```

```python
# ==============================================================================
# Task 2: Transfer Learning (Using MobileNetV2) for Comparison
# ==============================================================================
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import GlobalAveragePooling2D

# 1. Load the pre-trained MobileNetV2 model
base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(150, 150, 3))

# 2. Freeze the base model to use its learned features
base_model.trainable = False

# 3. Create the new model on top
transfer_model = Sequential([
    base_model,
    GlobalAveragePooling2D(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(1, activation='sigmoid')
])

# 4. Compile the transfer model
transfer_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# 5. Train the transfer model
print("🚀 Training Transfer Learning Model (MobileNetV2)...")
transfer_history = transfer_model.fit(
    train_generator,
    epochs=5, # 5 epochs is enough to see performance for transfer learning
    validation_data=val_generator
)

print("\n✅ Transfer Learning Model Training Completed!")
```

```python
# ==============================================================================
# Final Evaluation: Confusion Matrix & Classification Report
# ==============================================================================
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix

# 1. Get predictions from the transfer_model
# Note: Ensure test_generator is not shuffling for evaluation
test_generator.reset()
predictions = transfer_model.predict(test_generator)
y_pred = (predictions > 0.5).astype("int32")
y_true = test_generator.classes

# 2. Print Classification Report (Precision, Recall, F1-Score)
print("Classification Report:\n")
print(classification_report(y_true, y_pred, target_names=['NORMAL', 'PNEUMONIA']))

# 3. Plot Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['NORMAL', 'PNEUMONIA'], yticklabels=['NORMAL', 'PNEUMONIA'])
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix: MobileNetV2')
plt.show()
```
